```python
# 1. Install the Yahoo Finance library inside your Colab session
!pip install yfinance pandas

import os
import yfinance as yf
import pandas as pd

# 2. Define data collection function
def download_asset_data(ticker: str, start_date: str, end_date: str) -> pd.DataFrame:
    print(f"📡 Fetching data for ticker: {ticker} from {start_date} to {end_date}...")
    data = yf.download(ticker, start=start_date, end=end_date)
    print(f"✅ Successfully downloaded {len(data)} rows.")
    return data

# 3. Download Apple stock data
target_ticker = "AAPL"
df = download_asset_data(target_ticker, "2023-01-01", "2026-10-01")

# 4. Save to your local Colab file storage system
os.makedirs("data", exist_ok=True)
df.to_csv(f"data/{target_ticker}_historical.csv")
print(f"💾 File cleanly saved locally as: data/{target_ticker}_historical.csv")
```
